# Missing Values Treatment


```python
#Example: median imputation for numerical columns
# Students should review missingness before deciding whether to use this approach.

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
 if df[col].isnull().sum() > 0: df[col] = df[col].fillna(df[col].median())

print("Remaining missing values:", df.isnull().sum().sum())
```

```python
# Outlier investigation using IQR for Profit_INR
Q1 = df['Profit_INR'].quantile(0.25)
Q3 = df['Profit_INR'].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df['Profit_INR'] < (Q1 - 1.5 * IQR)) | (df['Profit_INR'] > (Q3 + 1.5 * IQR))]
print(f"\nNumber of profit outliers identified: {len(outliers)}")
```

```python
# IQR-based outlier summary for numerical columns

outlier_summary = []

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()
    outlier_summary.append({
        "Column": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": count,
        "Outlier_Percentage": round(count / len(df) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary).sort_values(
    "Outlier_Count", ascending=False
)

outlier_summary
```


#  Student-Designed Analyses
Conducting 3 specialized independent analyses:
1. Impact of Irrigation Methods on Water Efficiency.
2. Crop-wise Profitability Analysis.
3. Fertilizer Impact on Yield Across Seasons.

```python
print("--- ANALYSIS 1: Irrigation Method vs Water Efficiency ---")
irrigation_analysis = df.groupby('Irrigation_Method')['Water_Efficiency_t_per_1000m3'].mean().reset_index()
display(irrigation_analysis)

print("\n--- ANALYSIS 2: Top 5 Most Profitable Crops ---")
crop_profit = df.groupby('Crop')['Profit_INR'].mean().reset_index().sort_values(by='Profit_INR', ascending=False)
display(crop_profit.head())

print("\n--- ANALYSIS 3: Fertilizer Application vs Yield ---")
df['Fertilizer_Tier'] = pd.qcut(df['Fertilizer_kg_ha'], q=3, labels=['Low', 'Medium', 'High'])
fert_yield = df.groupby('Fertilizer_Tier')['Yield_Tonnes_Ha'].mean().reset_index()
display(fert_yield)
```
